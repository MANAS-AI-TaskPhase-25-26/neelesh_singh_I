```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("yeast.csv")
df.head(3)
```

```python
X = df.drop("name", axis=1)
y = df["name"]
```

```python
# Using again label encoding...
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_enc = le.fit_transform(y)
```

```python
# Printing encoding(not necessary but just to show...)
print("Class Encoding:")
for i, c in enumerate(le.classes_):
    print(i, "->", c)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y_enc, test_size=0.2, random_state=42, stratify=y)
```

```python
# Now Decision Tree model
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
```

```python
dt_pred = dt.predict(X_test)
```

```python
from sklearn.metrics import classification_report
print("\nDecision Tree Results")
print(classification_report(y_test, dt_pred))
```

```python
# Now Random Forest model
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=200,random_state=42,oob_score=True)
rf.fit(X_train, y_train)
```

```python
rf_pred = rf.predict(X_test)
```

```python
print("\nRandom Forest Results (Before fixing Imbalance..)")
print(classification_report(y_test, rf_pred))
print("OOB Score:", rf.oob_score_)
```

```python
# But we still have very low performance... SO checking the distribution
print("\nOriginal Class Distribution:")
print(y.value_counts())
```

Merging Rare Classes that have very low frequency...

```python
merge_map = {
    "VAC": "OTHER",
    "POX": "OTHER",
    "ERL": "OTHER",
    "ME1": "OTHER",
    "EXC": "OTHER"
}
y_new = y.replace(merge_map)
```

```python
le2 = LabelEncoder()
y_new_enc = le2.fit_transform(y_new)
```

```python
print("\nNew Class Encoding:")
for i, c in enumerate(le2.classes_):
    print(i, "->", c)
```

```python
print("\nNew Class Distribution:")
print(y_new.value_counts())
```

```python
X_train2, X_test2, y_train2, y_test2 = train_test_split(X, y_new_enc, test_size=0.2, random_state=42, stratify=y_new)
```

```python
rf_final = RandomForestClassifier(n_estimators=300,random_state=42,oob_score=True)
rf_final.fit(X_train2, y_train2)
```

```python
final_pred = rf_final.predict(X_test2)
```

```python
print("\nRandom Forest After Merging Rare Classes...")
print(classification_report(y_test2, final_pred))
print("Final OOB Score:", rf_final.oob_score_)
```
