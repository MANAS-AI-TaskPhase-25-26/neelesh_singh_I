```python
import pandas as pd
import numpy as np
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv("yeast.csv")
df.head(3)
```

```python
len(df["name"].unique()) # so there are 10 unique classification values....
```

```python
# Since model cannot read string data in name column, encoding it...
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["class"] = le.fit_transform(df["name"])
```

```python
df.head(3)
```

```python
X = df.drop(columns=["name","class"]).values
y = df["class"].values
```

```python
dt = DecisionTreeClassifier(criterion='gini',max_depth=None,min_samples_split=2,random_state=42)
```

```python
# Applying Cross-Validation..
from sklearn.model_selection import StratifiedKFold, cross_val_predict
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
y_pred = cross_val_predict(dt,X,y,cv=cv)
```

```python
y_prob = cross_val_predict(dt,X,y,cv=cv,method='predict_proba')
```

```python
# Evalutaing my model...
from sklearn.metrics import f1_score,recall_score,matthews_corrcoef,roc_auc_score
```

```python
f1 = f1_score(y,y_pred,average='weighted')
print("F1 Score: ",f1)
```

```python
recall = recall_score(y, y_pred, average="macro")
print("Recall: ", recall)
```

```python
mcc = matthews_corrcoef(y, y_pred)
print("Matthews Correlation Coefficient: ", mcc)
```

```python
auc = roc_auc_score(y, y_prob, multi_class="ovr")
print("ROC AUC: ",auc)
```
